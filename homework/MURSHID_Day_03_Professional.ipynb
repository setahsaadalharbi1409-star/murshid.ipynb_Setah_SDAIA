<div style="border-bottom:2px solid #087f83;padding-bottom:5px;margin-top:20px"><h2 style="color:#12314a;margin-bottom:0">02 | Validation logic and executable test cases</h2></div>

`validate_public_question(question)` enforces two rules: input must be a nonempty string after stripping spaces, and input length must not exceed 500 characters. It returns the normalized question when valid.

`run_error_examples()` tests three invalid values (`None`, whitespace-only text, and a 501-character string) and checks that the correct `ValueError` messages are raised. It then verifies one valid, whitespace-padded question.

| Test input | Expected behavior |
|:--|:--|
| `None` | Raise `ValueError`: nonempty-string requirement |
| Whitespace-only text | Raise `ValueError`: nonempty-string requirement |
| 501 characters | Raise `ValueError`: 500-character maximum |
| Valid text with outer spaces | Return trimmed text |

**Result structure:** `intentional_caught_errors`, `corrected_input`, and `checks`. A successful run reports the anticipated failures as handled, not as uncaught errors.


```python
# Deliberate, caught errors. These are examples to diagnose, not broken setup.
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()


def run_error_examples():
    cases = [(None, 'Enter a non-empty question string.'),
             ('   ', 'Enter a non-empty question string.'),
             ('x' * 501, 'Use at most 500 characters.')]
    rows = []
    for value, expected in cases:
        try:
            validate_public_question(value)
        except ValueError as error:
            assert str(error) == expected
            rows.append({'error_type': type(error).__name__, 'message': str(error), 'expected_rejection': True})
        else:
            raise AssertionError('Invalid input was accepted.')
    assert validate_public_question(' How can I check my application status? ') == 'How can I check my application status?'
    result = {'intentional_caught_errors': rows, 'corrected_input': 'accepted', 'checks': 'passed'}
    print(json.dumps(result, indent=2))
    return result



result = run_error_examples()
```
